```python
remove_cols = [
    'job_link',
    'last_processed_time',
    'got_summary',
    'got_ner',
    'is_being_worked',
    'first_seen'
]

df = df.drop(columns=remove_cols)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Country Distribution
plt.figure(figsize=(8, 5))
sns.histplot(df['search_country'], bins=20, kde=True)
plt.title('Applier Distribution by Country')
plt.xlabel('Country')
plt.ylabel('Number of Appliers')
plt.show()


# 2. Job Type Distribution
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='job_type')
plt.title('Applier Distribution by Job Type')
plt.xlabel('Job Type')
plt.ylabel('Number of Appliers')
plt.xticks(rotation=50)
plt.tick_params(axis='x', pad=0)
plt.tight_layout()
plt.show()


# 3. Job Level Distribution
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='job_level')
plt.title('Applier Distribution by Job Level')
plt.xlabel('Job Level')
plt.ylabel('Number of Appliers')
plt.xticks(rotation=50)
plt.tick_params(axis='x', pad=0)
plt.tight_layout()
plt.show()


# 4. Top 10 Job Titles
top_jobs = df['job_title'].value_counts().head(10)

top_jobs.plot(kind='bar')
plt.title('Top 10 Job Titles')
plt.xlabel('Job Title')
plt.ylabel('Number of Job Listings')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()


# 5. Top 10 Companies
top_companies = df['company'].value_counts().head(10)

top_companies.plot(kind='bar')
plt.title('Top 10 Companies')
plt.xlabel('Company')
plt.ylabel('Number of Job Listings')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()


# 6. Top 10 Search Positions
top_positions = df['search_position'].value_counts().head(10)

top_positions.plot(kind='bar')
plt.title('Top 10 Search Positions')
plt.xlabel('Search Position')
plt.ylabel('Number of Job Listings')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()


# 7. Top 10 Job Locations
top_locations = df['job_location'].value_counts().head(10)

top_locations.plot(kind='bar')
plt.title('Top 10 Job Locations')
plt.xlabel('Job Location')
plt.ylabel('Number of Job Listings')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()


# 8. Top 10 Search Cities
top_cities = df['search_city'].value_counts().head(10)

top_cities.plot(kind='bar')
plt.title('Top 10 Search Cities')
plt.xlabel('City')
plt.ylabel('Number of Job Listings')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Create list of available positions
positions = sorted(df['search_position'].dropna().unique().tolist())

print("Available Search Positions:")
print(positions)

# Get input from user
position = input(
    "\nEnter a search position: "
).strip()

# Check whether position is in the list
if position.lower() not in [p.lower() for p in positions]:
    print("Position not found. Please choose a position from the list.")

else:
    # Get the actual position name
    selected_position = next(
        p for p in positions if p.lower() == position.lower()
    )

    # Filter data
    position_df = df[df['search_position'] == selected_position]

    # Top 10 companies
    top_companies = position_df['company'].value_counts().head(10)

    # Bar chart
    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=top_companies.values,
        y=top_companies.index
    )

    plt.title(f'Top 10 Companies Hiring for {selected_position}')
    plt.xlabel('Number of Jobs')
    plt.ylabel('Company')
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create list of available countries
countries = sorted(
    df['search_country'].dropna().unique().tolist()
)

print("Available Countries:")
for country in countries:
    print(country)

# Get country from user
choice = input("\nEnter the country: ").strip()

# Check whether country is in the list
if choice.lower() not in [country.lower() for country in countries]:
    print("Country not found. Please choose a country from the list.")

else:
    # Get the actual country name
    selected_country = next(
        country for country in countries
        if country.lower() == choice.lower()
    )

    # Filter data
    country_df = df[
        df['search_country'] == selected_country
    ]

    # Top 10 companies
    top_companies = country_df['company'].value_counts().head(10)

    # Bar chart
    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=top_companies.values,
        y=top_companies.index
    )

    plt.title(f'Top 10 Companies in {selected_country}')
    plt.xlabel('Number of Jobs')
    plt.ylabel('Company')
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create list of available job levels
job_levels = sorted(
    df['job_level'].dropna().unique().tolist()
)

print("Available Job Levels:")
for level in job_levels:
    print(level)

# Get job level from user
choice = input("\nEnter the job level: ").strip()

# Check whether job level is in the list
if choice.lower() not in [level.lower() for level in job_levels]:
    print("Job level not found. Please choose a job level from the list.")

else:
    # Get the actual job level name
    selected_level = next(
        level for level in job_levels
        if level.lower() == choice.lower()
    )

    # Filter data
    level_df = df[
        df['job_level'] == selected_level
    ]

    # Top 10 job titles
    top_titles = level_df['job_title'].value_counts().head(10)

    # Bar chart
    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=top_titles.values,
        y=top_titles.index
    )

    plt.title(f'Top 10 Job Titles for {selected_level} Level')
    plt.xlabel('Number of Jobs')
    plt.ylabel('Job Title')
    plt.tight_layout()
    plt.show()
```
